# Background Problem

Data Analyst di **RevoFin**.
Dalam proyek analisis ini :
- Bakal ngegali **insight** dari data **portofolio pinjaman RevoFin**.
- Fokus **menemukan informasi** berharga yang bisa bantu **nyusun strategi bisnis dan ningkatin praktik pemberian pinjaman**.
- Analisisnya bakal nyentuh beberapa aspek, seperti **indikator kesehatan finansial, penilaian risiko, dan dinamika pelanggan**.
- Dengan ngulik ini, kamu bisa **nemuin pola dan tren yang berguna buat ngambil keputusan** yang lebih tepat dan **nge-optimalin layanan pinjaman**.

Target audience **Risk Management Team** dan **Executive Management**


Objectives of the Analysis :
- Menampilkan **jumlah pinjaman yang masih outstanding** dalam portofolio.
- **Menyoroti metrik risiko yang terkait** dengan portofolio pinjaman.
- Memberikan **analisis mendalam** terkait pinjaman outstanding dan metrik risikonya berdasarkan:  
1.   Karakteristik pinjaman (misalnya jenis, tenor, suku bunga).
2.   Atribut pelanggan (misalnya tingkat pendapatan, status pekerjaan, kepemilikan rumah).
- **Menunjukkan kemampuan** kamu dalam menggunakan SQL.
